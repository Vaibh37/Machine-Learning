before training i sometimes need to change my input data

StandardScaler helps with numeric features

## StandardScaler

it learns the mean and spread of each numeric feature

then it changes the values to a new scale

the transformed values can be negative and that is fine


```python
from sklearn.preprocessing import StandardScaler

X = [[10, 100], [20, 200], [30, 300]]

scaler = StandardScaler()
scaled = scaler.fit_transform(X)

print(scaled)
print("mean", scaler.mean_)
```

## OneHotEncoder

a category like red is not the same as a number

the encoder makes a column for each category

one column is on for each input category


```python
from sklearn.preprocessing import OneHotEncoder

colors = [["red"], ["blue"], ["red"], ["green"]]

encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(colors)

print(encoder.get_feature_names_out())
print(encoded)
```

## important when i use train and test data

learn the preprocessing rules from training data only

use transform on test data and new data

do not learn preprocessing from the test answers


```python
new_colors = [["blue"], ["green"]]
print(encoder.transform(new_colors))
```
