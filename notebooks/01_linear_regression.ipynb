# linear regression

i use linear regression when i want to predict a number

this first dataset is simple on purpose

X is the input and y is the answer


```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

X = [[1], [2], [3], [4], [5]]
y = [100, 200, 300, 400, 500]

print(X)
print(y)
```

## train

fit looks at X and y and learns a line

i do not tell the model the slope by hand


```python
model = LinearRegression()
model.fit(X, y)

print("slope", model.coef_)
print("intercept", model.intercept_)
```

## predict a new number

the model has only seen inputs from 1 to 5

6 is a new input

this is only a tiny example


```python
prediction = model.predict([[6]])
print("prediction for 6", prediction)
```

## check the error

mae means mean absolute error

mse means mean squared error

r2 is a score that compares the model with a simple baseline

here we check the same data used for training

this does not prove the model works well on new data


```python
training_predictions = model.predict(X)

print("MAE", mean_absolute_error(y, training_predictions))
print("MSE", mean_squared_error(y, training_predictions))
print("R2", r2_score(y, training_predictions))
```
