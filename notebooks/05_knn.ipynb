# knn

knn means k nearest neighbours

it predicts by looking at training points that are close to the new point

for one number i can use absolute distance

it checks the distances and uses the nearest k points


```python
from sklearn.neighbors import KNeighborsClassifier

X = [[1], [2], [4], [8]]
y = [0, 0, 1, 1]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

print("for 3", model.predict([[3]]))
print("for 7", model.predict([[7]]))
```

## i was stuck here

knn does not check only the left side or only the right side

it looks at all training points

for a new value of 3

distance to 1 is 2

distance to 2 is 1

distance to 4 is 1

distance to 8 is 5

the nearest three are 2 and 4 and 1

their labels are 0 and 1 and 0

so the predicted class is 0


```python
new_value = 3
distances = [(point[0], abs(point[0] - new_value)) for point in X]
distances = sorted(distances, key=lambda item: item[1])

print("point and distance")
print(distances)
print("nearest three")
print(distances[:3])
```

## what k means

k is the number of nearest training points used for the vote

it is not the number of classes

if k is 5 then the model needs at least five training rows


```python
X_more = [[1], [2], [4], [6], [8], [10]]
y_more = [0, 0, 1, 1, 1, 1]

model_five = KNeighborsClassifier(n_neighbors=5)
model_five.fit(X_more, y_more)

print("for 3 with k equal to 5", model_five.predict([[3]]))
```
