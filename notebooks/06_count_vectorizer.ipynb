# countvectorizer

this tool turns text into word counts

it does not decide spam or not spam

it only prepares features for a model


```python
from sklearn.feature_extraction.text import CountVectorizer

messages = [
    "python is fun",
    "python is powerful",
    "machine learning is fun",
    "python python learning",
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(messages)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

## how the numbers work

first it learns a list of unique words

the feature names come out in alphabetical order in this example

each word has a fixed column

each row is one message

the number tells me how many times that word appears

the last message has python twice so its python column has 2


## three methods

fit_transform learns the words and counts them for training text

get_feature_names_out shows the words in column order

toarray lets me see the whole number matrix

CountVectorizer normally gives me a sparse matrix

a sparse matrix saves space by not storing every zero as a full number


```python
print("type of X", type(X))
print("shape", X.shape)
print("words", vectorizer.get_feature_names_out())
```

## change new text without learning the words again

transform uses the vocabulary from training

a new word that is not in that vocabulary gets ignored


```python
new_message = ["python is python"]
new_X = vectorizer.transform(new_message)

print(new_X.toarray())
```
