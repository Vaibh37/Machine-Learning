# classification metrics

accuracy alone can hide a problem

i also want to know which kinds of mistakes the model makes

in this example 1 is the positive class


```python
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    precision_score,
    recall_score,
)

y_true = [1, 0, 1, 1, 0, 1]
y_pred = [1, 0, 0, 1, 1, 1]

print("accuracy", accuracy_score(y_true, y_pred))
print("confusion matrix")
print(confusion_matrix(y_true, y_pred))
```

## reading the confusion matrix

for labels 0 then 1 the rows are the real classes

the columns are the predicted classes

top left is true negative

top right is false positive

bottom left is false negative

bottom right is true positive

our result is

1 1

1 3

so there is one false positive and one false negative


```python
matrix = confusion_matrix(y_true, y_pred)
tn, fp, fn, tp = matrix.ravel()

print("true negative", tn)
print("false positive", fp)
print("false negative", fn)
print("true positive", tp)
```

## precision and recall

precision asks

out of all predicted positives how many were correct

recall asks

out of all real positives how many did we find

the library calculates both for me


```python
print("precision", precision_score(y_true, y_pred))
print("recall", recall_score(y_true, y_pred))
```

## F1

F1 gives one score that combines precision and recall
