# naive bayes

this is my small spam classifier

1 means spam

0 means not spam

the model learns word patterns from labelled emails

it uses probability but i do not need to calculate every step by hand

this is a toy dataset and not a safe real spam filter


## first i learned it with manual word counts

in the next example

first column counts free

second column counts meeting


```python
from sklearn.naive_bayes import MultinomialNB

X = [[2, 0], [1, 0], [0, 2], [0, 1]]
y = [1, 1, 0, 0]

model = MultinomialNB()
model.fit(X, y)

print("class", model.predict([[1, 0]]))
print("class order", model.classes_)
print("probabilities", model.predict_proba([[1, 0]]))
```

## now use real text

CountVectorizer turns email text into word counts

MultinomialNB learns which word counts go with each class

i do not tell the model that free is spam

the labels help it learn that pattern from the examples


```python
from sklearn.feature_extraction.text import CountVectorizer

emails = [
    "free prize now",
    "claim your free gift",
    "team meeting today",
    "project meeting tomorrow",
]

y = [1, 1, 0, 0]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(emails)

print("words", vectorizer.get_feature_names_out())
print("word counts")
print(X.toarray())
```

## train

X contains the word counts

y contains the right label for each email

fit learns from both together


```python
model = MultinomialNB()
model.fit(X, y)
```

## a new email

the new email is still text

first i use transform to turn it into numbers

transform keeps the same vocabulary and same column order

then the model can predict


```python
new_email = ["free gift"]
new_X = vectorizer.transform(new_email)

print("new word counts", new_X.toarray())
print("class", model.predict(new_X))
print("class order", model.classes_)
print("class probabilities", model.predict_proba(new_X))
```

## a different email

meeting is linked with not spam in this tiny training set

check what the model predicts


```python
new_email = ["team meeting"]
new_X = vectorizer.transform(new_email)
print("class", model.predict(new_X))
```
